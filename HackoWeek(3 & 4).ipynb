```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

data = pd.read_csv('E-commerce_cosmetic_dataset_cleaned.csv')

data.head()
```

```python
data.drop(['Unnamed: 0'], axis=1, inplace = True)
data.head()
```

```python
data.describe()
```

```python
all_brands = data['brand'].unique()
all_brands
```

```python
all_brands.sort()
all_brands
```

```python
rating_by_brand = (data.groupby('brand', group_keys=True)[['rating']].mean())['rating']
rating_by_brand
```

```python
plt.barh(all_brands[0:21], rating_by_brand[0:21])
plt.title('Mean rating by brand')
plt.xlabel('Ratings (0-5)')
plt.ylabel('Brands')
plt.show()
```

```python
all_categories = data['category'].unique()

all_categories.sort()
all_categories
```

```python
price_by_category = (data.groupby('category', group_keys=True)[['price']].mean())['price']
price_by_category
```

```python
plt.bar(all_categories, price_by_category)
plt.title('Mean price by subcategory')
plt.xlabel('Categories')
plt.ylabel('Price (in INR)')
plt.show()
```
