```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split, cross_val_score, KFold
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

df = pd.read_csv('Dummy Data HSS.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
le = LabelEncoder()

df['Influencer'] = le.fit_transform(df['Influencer'])
df.head()
```

```python
#Train-Test Split & Cross-Validation
X = df[['TV', 'Radio', 'Social Media', 'Influencer']]

y = df['Sales']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0)

model1 = LinearRegression()
scores1 = cross_val_score(model1, X, y, cv=5)
print(scores1)

model2 = Ridge(alpha=1.0)
scores2 = cross_val_score(model2, X, y, cv=5)
print("\n", scores2)

model3 = RandomForestRegressor(max_depth=2, random_state=42)
scores3 = cross_val_score(model3, X, y, cv=5)
print("\n", scores3)
```

```python
print("Linear Regression")
print(f"Accuracy: {scores1.mean().round(3)}, Standard Deviation: {scores1.std().round()}\n\n")

print("Ridge Regression")
print(f"Accuracy: {scores2.mean().round(3)}, Standard Deviation: {scores2.std().round()}\n\n")

print("Random Forest Classification")
print(f"Accuracy: {scores3.mean().round(3)}, Standard Deviation: {scores3.std().round()}\n\n")
```

```python
model1.fit(X_train, y_train)
y_pred1 = model1.predict(X_test)

model2.fit(X_train, y_train)
y_pred2 = model2.predict(X_test)

model3.fit(X_train, y_train)
y_pred3 = model3.predict(X_test)

bins = [0, 100.0, 200.0, 300.0, 400.0]
y_test_binned = pd.cut(y_test, bins = bins)
y_pred1_binned = pd.cut(y_pred1, bins = bins)
y_pred2_binned = pd.cut(y_pred2, bins = bins)
y_pred3_binned = pd.cut(y_pred3, bins = bins)
```

```python
y_test_binned
```

```python
cm1 = confusion_matrix(y_test_binned, y_pred1_binned)
cm2 = confusion_matrix(y_test_binned, y_pred2_binned)
cm3 = confusion_matrix(y_test_binned, y_pred3_binned)

disp1 = ConfusionMatrixDisplay(confusion_matrix=cm1, display_labels=model1.classes_)
disp2 = ConfusionMatrixDisplay(confusion_matrix=cm2, display_labels=model2.classes_)
disp3 = ConfusionMatrixDisplay(confusion_matrix=cm3, display_labels=model3.classes_)

disp1.plot()
disp2.plot()
disp3.plot()
```
